### Step 3: Download Competition Dataset

```python
# Download latest version of the competition data
path = kagglehub.competition_download('rsna-pneumonia-detection-challenge')

print("Path to competition files:", path)
```

### Step 4: Inspect Downloaded Files

```python
import os
import pandas as pd

print("Contents of dataset directory:")
for item in os.listdir(path):
    item_path = os.path.join(path, item)
    if os.path.isdir(item_path):
        count = len(os.listdir(item_path))
        print(f"  [DIR]  {item} ({count} files)")
    else:
        size_mb = os.path.getsize(item_path) / (1024 * 1024)
        print(f"  [FILE] {item} ({size_mb:.2f} MB)")
```

### Step 5: Read Labels CSV & Preview First Rows

```python
train_labels_file = os.path.join(path, "stage_2_train_labels.csv")
if os.path.exists(train_labels_file):
    df_labels = pd.read_csv(train_labels_file)
    print("Shape:", df_labels.shape)
    display(df_labels.head())
else:
    print("Labels file not found yet.")
```
